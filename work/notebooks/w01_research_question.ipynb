## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
## 2. The question

### Research question

**Which pages should be reviewed first for a possible content refresh or improvement, based
on observable search and engagement signals?**

### Decision

The decision is which pages should receive attention first when the content team has limited
review capacity.

### Who acts on the output?

A content or SEO reviewer would use the ranked list to select pages for manual review and then
decide whether to refresh, improve, protect, monitor, or take another appropriate action.

### Cost of a wrong call

A false positive can waste an editor's or SEO reviewer's time on a page that does not need
attention. A false negative can cause a potentially useful page to be missed. Therefore the
ranking should be treated as decision support rather than an automatic action.

### Why data or ML can help

A simple rule can identify obvious cases, but the dataset contains several signals at once,
including impressions, sessions, CTR, position, content age, freshness, and engagement.
A model or scoring method may help combine these signals when their relationships are too
messy to express with a few simple rules.

### Task type and metric

This is a **ranking/scoring** problem. The output will be a priority score that can be used
to rank pages for review. A suitable evaluation metric is **Precision@K**, because the real
decision involves reviewing a limited number of pages first.
```
